```python
import pandas as pd
from pathlib import Path
from IPython.display import display, HTML
# Declare sources
data_dir = Path("Data")
baked_data_dir = Path("BakedData")
SourceA = "Source A"
SourceB = "Source B"
```

```python
# Bake Source A
pd.read_excel(data_dir / f"{SourceA}.xlsx", header=[0, 1]).to_csv(baked_data_dir / f"{SourceA}.csv", index=False)
#pd.read_excel(data_dir / f"{SourceA}.xlsx", header=[0, 1]).to_csv(baked_data_dir / f"{SourceA}.csv")
print("Source A baked")
```

```python
# Bake Source B
pd.read_excel(data_dir / f"{SourceB}.xlsx", header=[0, 1]).to_csv(baked_data_dir / f"{SourceB}.csv", index = False)
print("Source B baked")
```

```python
# Load Source A
df_a = pd.read_csv(baked_data_dir / f"{SourceA}.csv")
# Bake units in my case as first row
df_a_units = df_a.iloc[0:1]
#print(df_a_units)
print(f"Loaded: {SourceA}")
```

```python
# Load Source B
df_b = pd.read_csv(baked_data_dir / f"{SourceB}.csv")
# Bake units in my case as first row
df_b_units = df_b.iloc[0:1]
#print(df_b_units)
print(f"Loaded: {SourceB}")
```

```python
headers_a = list(df_a.columns)
headers_b = list(df_b.columns)
#print(headers_a)
#print(headers_b)
match = 1 if headers_a == headers_b else 0

if match:
    print("Validation success — headers match")
else:
    print("Validation failed — headers differ")
    diff_a = set(headers_a) - set(headers_b)
    diff_b = set(headers_b) - set(headers_a)
    if diff_a:
        print(f"  Only in Source A: {diff_a}")
    if diff_b:
        print(f"  Only in Source B: {diff_b}")
```

```python
#Bake MergedData
merged = df_a.merge(df_b, on="name", suffixes=("_A", "_B"))
merged_units = merged.iloc[0:1]
#show(merged,"combinated", start= 5, rows = 20, persistence_data= merged_units)
```
